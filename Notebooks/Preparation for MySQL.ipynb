```python
# Function to extract minimum and maximum age from the 'Age' column
def extract_age_range(age_str):
    '''Extracts minimum and maximum age from a given age string.'''

    if pd.isna(age_str):
        return None, None
    
    age_str = str(age_str).lower()
    
    # Normalizing time units to years
    def to_years(val, unit):
        '''Converts a value with a given time unit to years.'''
        val = float(val)
        if 'month' in unit: return round(val / 12, 2)
        if 'week' in unit: return round(val / 52, 2)
        if 'day' in unit: return round(val / 365, 2)
        return val

    # Extracting all age values and their corresponding units
    matches = re.findall(r'(\d+)\s+(year|month|week|day)', age_str)
    
    min_age, max_age = None, None
    
    if matches:
        years_list = [to_years(val, unit) for val, unit in matches]
        
        if "and older" in age_str:
            min_age = years_list[0]
            max_age = 120.0 # to represent "and older"
        elif "up to" in age_str:
            min_age = 0.0
            max_age = years_list[0]
        elif "to" in age_str and len(years_list) >= 2:
            min_age = years_list[0]
            max_age = years_list[1]
        elif len(years_list) == 1:
            min_age = years_list[0]
            max_age = None
    else:
        # for cases where age is described in terms of categories (e.g., "child", "adult", "older adult")
        if "child" in age_str:
            min_age = 0
            if "older adult" in age_str: max_age = 120
            elif "adult" in age_str: max_age = 64
            else: max_age = 17
        elif "older adult" in age_str:
            min_age = 65
            max_age = 120
        elif "adult" in age_str:
            min_age = 18
            max_age = 64
            
    return min_age, max_age

# Aplying the function to extract minimum and maximum age and creating new columns
df_studies[['minimum_age', 'maximum_age']] = df_studies['Age'].apply(lambda x: pd.Series(extract_age_range(x)))

```

```python
# Selection of columns for the locations table
df_locations = df[['Rank','Locations']].copy()

# Addressing the issue of multiple locations in a single string by splitting them into lists
df_locations['Locations'] = df_locations['Locations'].str.split('|')

# Exploding the 'Locations' column to create a new row for each location
df_locations = df_locations.explode('Locations')
```

```python
# Function to split location details into separate columns
def split_location_details(loc_str):
    if pd.isna(loc_str) or loc_str == '':
        return None, None, None, None
    
    # Cleaning whitespaces and splitting the location string by commas
    parts = [p.strip() for p in str(loc_str).split(',')]
    n = len(parts)
    
    facility, city, state, country = None, None, None, None
    
    if n >= 2:
        # Country always appears last
        country = parts[-1]
        
        # Defining a list of countries that typically include state information in the location string
        countries_with_states = ['united states', 'usa', 'us', 'canada', 'brazil', 'australia', 'mexico', 'colombia']
        
        if country.lower() in countries_with_states and n >= 3:
            # Logic for countries with states
            state = parts[-2]
            city = parts[-3]
            # All that remains on the left is the Facility
            facility = ", ".join(parts[:-3]) if n > 3 else None
        else:
            # Logic for Europe/Rest of the World
            city = parts[-2]
            # All that remains on the left is the Facility
            facility = ", ".join(parts[:-2]) if n > 2 else None
            state = None 

    # If the facility is empty, the first is the city
    if not facility and n == 2:
        city, country = parts[0], parts[1]
        
    return facility, city, state, country

# Applying the function to split location details into separate columns
df_locations[['facility', 'city', 'state', 'country']] = df_locations['Locations'].apply(
    lambda x: pd.Series(split_location_details(x))
)
```

```python
# Function to extract specific values from the 'Study Designs' string
def extract_design_attribute(text, attribute):
    '''
    Extracts specific attributes from a concatenated string.
    
    The 'Study Designs' column in the raw dataset contains multiple key-value pairs separated by pipes (|). 
    This function parses the string, identifies the requested attribute (e.g., 'Masking:', 'Allocation:'), and returns the corresponding clean value.
    
    Parameters:
    text (str): The raw string from the 'Study Designs' column.
    attribute (str): The label to search for within the string.
    
    Returns:
    str: The extracted attribute value, or None if the attribute is not found or the input text is null.
    '''
    if pd.isna(text):
        return None
    # Split the string by the pipe '|' character
    parts = text.split('|')
    for part in parts:
        if attribute in part:
            # Remove the label and return the clean value
            return part.replace(attribute, '').strip()
    return None

# Applying the extraction to create 4 new clean columns
df_design['allocation'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Allocation:'))
df_design['intervention_model'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Intervention Model:'))
df_design['observational_model'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Observational Model:'))
df_design['time_perspective'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Time Perspective:'))
df_design['masking'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Masking:'))
df_design['primary_purpose'] = df_design['Study Designs'].apply(lambda x: extract_design_attribute(x, 'Primary Purpose:'))

```

```python
# Selecting columns for the sponsors table
df_sponsors_raw = df[['Rank', 'Sponsor/Collaborators', 'Funded Bys']].copy()

# Renaming columns to match the database schema
df_sponsors_raw = df_sponsors_raw.rename(columns={'Funded Bys': 'agency_class', 'Rank':'study_id'})

# Function to separate agencies and define who is Lead or Collaborator
# The first agency in the list is considered the Lead, and the others are Collaborators
def process_sponsors(row):
    '''
    This function takes a row of the dataframe, separates the sponsors/collaborators and their classes,
    and returns a list of dictionaries with the agency name, class, and role (Lead or Collaborator).
    '''
    
    agencies = str(row['Sponsor/Collaborators']).split('|')
    classes = str(row['agency_class']).split('|')
    
    results = []
        
    for i, name in enumerate(agencies):
        # The first in the list (index 0) is the Lead, the others are Collaborator
        role = 'Lead' if i == 0 else 'Collaborator'
        
        # Find the corresponding class (if it exists in the same order)
        # If no class is found for all, mark as 'Other' or 'Unknown'
        current_class = classes[i].strip() if i < len(classes) else "Other"
        
        results.append({
            'agency': name.strip(),
            'agency_class': current_class,
            'lead_or_collaborator': role
        })
    return results

# Creating a temporary column with the list of dictionaries
df_sponsors_raw['temp_list'] = df_sponsors_raw.apply(process_sponsors, axis=1)

# Separating the list of dictionaries into individual rows
df_sponsors_final = df_sponsors_raw.explode('temp_list')

# Converting the list of dictionaries into separate columns and cleaning the dataframe
df_sponsors_final = pd.concat([
    df_sponsors_final[['study_id']], 
    df_sponsors_final['temp_list'].apply(pd.Series)
], axis=1)

# Final cleaning
df_sponsors_final = df_sponsors_final[df_sponsors_final['agency'] != 'nan']
df_sponsors_final = df_sponsors_final.dropna(subset=['agency'])
```

```python

# Removing empty rows
df_outcome = df_outcome.dropna(subset=['Outcome Measures'])
```

```python
# Separating outcome measures by '|' and expanding into multiple rows
df_outcome['Outcome Measures'] = df_outcome['Outcome Measures'].str.split('|')
df_outcome = df_outcome.explode('Outcome Measures')

# Cleaning whitespace from outcome measures and dropping the original column
df_outcome['outcome_measure_clean'] = df_outcome['Outcome Measures'].str.strip()
df_outcome.drop(columns=['Outcome Measures'], inplace=True)
```

```python
# Function to split the outcome measure into type, measure, time frame, and description
def split_outcome_logic_v2(text):
    '''Splits the outcome measure text into structured components: outcome type, measure, time frame, and description.'''
    if pd.isna(text):
        return "Not Specified", None, "Not Specified", "No description provided"
    
    # Using a comprehensive regex to capture all time patterns
    time_pattern = r'((?:at|after|from|within|and)\s+(?:\d+[\s-]*months?|\d+[\s-]*weeks?|\d+[\s-]*days?|\d+[\s-]*years?|birth|baseline))|(\d+[\s-]*months?|\d+[\s-]*weeks?|\d+[\s-]*days?|\d+[\s-]*years?)'
    
    # Find all time references
    matches = re.findall(time_pattern, text, re.IGNORECASE)
    # cleaning the matches to get a list of time frames
    time_list = [max(m) for m in matches if any(m)] 
    
    time_frame = ", ".join(time_list) if time_list else "Not Specified"
    
    # Separating the measure from the time frame by removing the time references from the original text
    measure = text
    for t in time_list:
        measure = measure.replace(t, "")
    
    # Stripping extra spaces and punctuation from the measure
    measure = measure.strip(" ,.;")
    
    # Since the text is very free-form, the original text was set as the measure and the description pointed to it
    description = "Detailed in measure" if len(measure) > 0 else "No description provided"
    
    return "Not Specified", measure, time_frame, description # there is no clear way to determine the outcome type from the text, so it will be returned as "Not Specified"

# Applying the function to split the outcome measures into structured components
df_outcome[['outcome_type', 'measure', 'time_frame', 'description']] = df_outcome['outcome_measure_clean'].apply(
    lambda x: pd.Series(split_outcome_logic_v2(x))
)
```

```python
# Normalization of the 'measure' column to reduce duplicates and improve analysis
# Converting to lowercase to standardize the text
df_outcome['measure'] = df_outcome['measure'].str.lower()

# Removing special characters that could cause duplicates (e.g., hyphens, multiple spaces)
df_outcome['measure'] = df_outcome['measure'].str.replace('-', ' ', regex=False)
df_outcome['measure'] = df_outcome['measure'].str.replace(r'\s+', ' ', regex=True).str.strip()

# Dropping duplicates based on the normalized 'measure' 
# Rows are considered duplicates if the study_id, measure and time_frame are the same
df_outcome = df_outcome.drop_duplicates(subset=['study_id', 'measure', 'time_frame'])

# Capitalizing the first letter of each measure
df_outcome['measure'] = df_outcome['measure'].str.capitalize()
```

```python
# Verifying the number of duplicates 
outcome_duplicates = df_outcome.duplicated(subset=['study_id', 'measure', 'time_frame']).sum()

print(f"There are {outcome_duplicates} duplicates in the outcome measures.")
```

```python
# Normalization of the 'time_frame' column to have identical time frames represented in the same way
# Initial standardization: Convert to lowercase, replace hyphens with spaces, and ensure a space between numbers and units (e.g., "6months" to "6 months")
df_outcome['time_frame'] = (df_outcome['time_frame']
                            .str.lower()
                            .str.replace('-', ' ', regex=False)
                            .str.replace(r'(\d+)([a-z])', r'\1 \2', regex=True))

# Removing the word "at" from time frames, (except when it is followed by "birth" or "baseline")
df_outcome['time_frame'] = df_outcome['time_frame'].str.replace(r'\b(at|within)\s+(?!birth|baseline)', '', regex=True)

# Converting plural time units to singular
df_outcome['time_frame'] = df_outcome['time_frame'].str.replace(r'\b(days|weeks|months|years)\b', lambda m: m.group(1)[:-1], regex=True)

# Aplying pluralization except when the number is 1
def pluralize(match):
    num = match.group(1)
    unit = match.group(2)
    if num == "1":
        return f"{num} {unit}"
    else:
        return f"{num} {unit}s"

df_outcome['time_frame'] = df_outcome['time_frame'].str.replace(r'\b(\d+)\s+(day|week|month|year)\b', pluralize, regex=True)

# Final cleaning and capitalization
df_outcome['time_frame'] = df_outcome['time_frame'].str.replace(r'\s+', ' ', regex=True).str.strip().str.capitalize()
```
